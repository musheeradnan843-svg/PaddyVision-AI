```python
# ==================================
# Inspect first-level dataset files
# ==================================

def inspect_folder(folder_path):
    print(f"\n{'=' * 60}")
    print(f"DATASET: {folder_path.name}")
    print(f"{'=' * 60}")

    if not folder_path.exists():
        print("Folder not found!")
        return

    items = list(folder_path.iterdir())

    folders = [item for item in items if item.is_dir()]
    files = [item for item in items if item.is_file()]

    print(f"Folders : {len(folders)}")
    print(f"Files   : {len(files)}")

    if folders:
        print("\nFolders:")
        for folder in sorted(folders):
            print(f"  📁 {folder.name}")

    if files:
        print("\nFiles:")
        for file in sorted(files):
            size_kb = file.stat().st_size / 1024
            print(f"  📄 {file.name}  ({size_kb:.2f} KB)")


inspect_folder(WEEDY_DIR)
inspect_folder(UAV_DIR)
inspect_folder(ICAR_DIR)
inspect_folder(HISTORICAL_DIR)
```

```python
# ============================
# File extension distribution
# ============================

def get_file_extensions(folder_path):
    extensions = {}

    for file in folder_path.rglob("*"):
        if file.is_file():
            ext = file.suffix.lower() or "[no extension]"
            extensions[ext] = extensions.get(ext, 0) + 1

    return extensions


for name, folder in {
    "Weedy Rice": WEEDY_DIR,
    "Rice UAV": UAV_DIR,
    "ICAR Rice": ICAR_DIR,
    "Historical Rice": HISTORICAL_DIR
}.items():

    print(f"\n{name}")
    print("-" * 40)

    extensions = get_file_extensions(folder)

    for ext, count in sorted(extensions.items()):
        print(f"{ext:15} : {count}")
```

```python
import pandas as pd
from pathlib import Path

# 1. Sheet-wise DataFrames store karne ke liye dictionary
master_sheets = {
    "Timeseries": [],
    "Weather data": [],
    "Treatments": [],
    "Timeline field activities": []
}

# 2. Sabhi 93 files par loop chalana aur metadata append karna
for file_path in historical_files:
    try:
        excel = pd.ExcelFile(file_path)
        file_name = file_path.name
        region = file_path.parent.name  # Central_India_Region ya Gangetic_Plain_Region
        
        for sheet in excel.sheet_names:
            if sheet in master_sheets:
                df = pd.read_excel(file_path, sheet_name=sheet)
                
                # Source tracking ke liye metadata columns add kar rahe hain
                df.insert(0, "Source_File", file_name)
                df.insert(1, "Region", region)
                
                master_sheets[sheet].append(df)
    except Exception as e:
        print(f"Error reading {file_path.name}: {e}")

# 3. Har sheet type ko ek single Master DataFrame mein merge karna
combined_datasets = {}
print("--- Consolidation Summary ---")
for sheet_name, df_list in master_sheets.items():
    if df_list:
        # Concatenate keeping all columns (union of columns across files)
        combined_df = pd.concat(df_list, ignore_index=True)
        combined_datasets[sheet_name] = combined_df
        print(f"Sheet: '{sheet_name}' -> Total Rows: {combined_df.shape[0]}, Total Columns: {combined_df.shape[1]}")

# 4. Cleaned/Merged data ko CSVs mein save karna
processed_dir = PROJECT_ROOT / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

for sheet_name, df in combined_datasets.items():
    clean_name = sheet_name.lower().replace(" ", "_")
    output_path = processed_dir / f"historical_{clean_name}.csv"
    df.to_csv(output_path, index=False)
    print(f"Saved: {output_path.name}")
```

```python
from collections import defaultdict

multispectral_dir = WEEDY_DIR / "Multispectral"

# Har image ID ke available bands store karenge
image_bands = defaultdict(set)

for tif_path in multispectral_dir.glob("*.TIF"):
    filename = tif_path.stem

    # Last part G/R/RE/NIR hoga
    band = filename.split("_")[-1]

    # Band ko filename se remove karke image ID
    image_id = filename.rsplit("_", 1)[0]

    if band in {"G", "R", "RE", "NIR"}:
        image_bands[image_id].add(band)

# Summary
required_bands = {"G", "R", "RE", "NIR"}

complete_ids = []
incomplete_ids = []

for image_id, bands in image_bands.items():
    missing = required_bands - bands

    if not missing:
        complete_ids.append(image_id)
    else:
        incomplete_ids.append((image_id, sorted(bands), sorted(missing)))

print("Total unique image IDs:", len(image_bands))
print("Complete 4-band sets:", len(complete_ids))
print("Incomplete sets:", len(incomplete_ids))
```

```python
import rasterio
import numpy as np

# Folder se pehli Green band file auto-pick karein (dono extension check karke)
g_files = list(ms_dir.glob("*_G.TIF")) + list(ms_dir.glob("*_G.tif"))

if g_files:
    sample_g_path = g_files[0]
    # Automatic sample_base extraction
    base_name = sample_g_path.name.replace("_G.TIF", "").replace("_G.tif", "")
    print(f"Found Sample: {base_name}")

    def read_band(suffix):
        path = ms_dir / f"{base_name}{suffix}"
        if not path.exists():
            path = ms_dir / f"{base_name}{suffix.lower()}"
        with rasterio.open(path) as src:
            return src.read(1).astype(np.float32)

    G = read_band("_G.TIF")
    R = read_band("_R.TIF")
    RE = read_band("_RE.TIF")
    NIR = read_band("_NIR.TIF")

    print("Successfully loaded all 4 bands!")
    print(f"Band Shape: {G.shape}")
else:
    print("Multispectral folder me koi file nahi mili!")
```

```python
# NDVI
ndvi_denominator = NIR + R

NDVI = np.divide(
    NIR - R,
    ndvi_denominator,
    out=np.zeros_like(NIR),
    where=ndvi_denominator != 0
)

# NDRE
ndre_denominator = NIR + RE

NDRE = np.divide(
    NIR - RE,
    ndre_denominator,
    out=np.zeros_like(NIR),
    where=ndre_denominator != 0
)

print("NDVI")
print("Shape:", NDVI.shape)
print("Min:", NDVI.min())
print("Max:", NDVI.max())
print("Mean:", NDVI.mean())

print("\nNDRE")
print("Shape:", NDRE.shape)
print("Min:", NDRE.min())
print("Max:", NDRE.max())
print("Mean:", NDRE.mean())
```

```python
from pathlib import Path

ndvi_dir = PROJECT_ROOT / "data" / "processed" / "ndvi"
ndre_dir = PROJECT_ROOT / "data" / "processed" / "ndre"

ndvi_dir.mkdir(parents=True, exist_ok=True)
ndre_dir.mkdir(parents=True, exist_ok=True)

# Sabhi G-band files
g_files = list(ms_dir.glob("*_G.TIF")) + list(ms_dir.glob("*_G.tif"))

processed = 0
skipped = 0

for g_path in g_files:

    base_name = g_path.name.rsplit("_G", 1)[0]

    r_path = ms_dir / f"{base_name}_R.TIF"
    re_path = ms_dir / f"{base_name}_RE.TIF"
    nir_path = ms_dir / f"{base_name}_NIR.TIF"

    # Agar lowercase extension ho
    if not r_path.exists():
        r_path = ms_dir / f"{base_name}_R.tif"
    if not re_path.exists():
        re_path = ms_dir / f"{base_name}_RE.tif"
    if not nir_path.exists():
        nir_path = ms_dir / f"{base_name}_NIR.tif"

    if not (r_path.exists() and re_path.exists() and nir_path.exists()):
        skipped += 1
        continue

    with rasterio.open(r_path) as src:
        R = src.read(1).astype(np.float32)

    with rasterio.open(re_path) as src:
        RE = src.read(1).astype(np.float32)

    with rasterio.open(nir_path) as src:
        NIR = src.read(1).astype(np.float32)

    # NDVI
    denominator = NIR + R
    NDVI = np.divide(
        NIR - R,
        denominator,
        out=np.zeros_like(NIR),
        where=denominator != 0
    )

    # NDRE
    denominator = NIR + RE
    NDRE = np.divide(
        NIR - RE,
        denominator,
        out=np.zeros_like(NIR),
        where=denominator != 0
    )

    # Save as NumPy arrays
    np.save(ndvi_dir / f"{base_name}_NDVI.npy", NDVI)
    np.save(ndre_dir / f"{base_name}_NDRE.npy", NDRE)

    processed += 1

print("Processing complete!")
print("Processed:", processed)
print("Skipped:", skipped)
```
